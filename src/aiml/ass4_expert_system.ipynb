```python
rules = [
    (["fever", "cough"], "flu"),
    (["flu", "body_pain"], "see_doctor"),
    (["sneezing", "runny_nose"], "cold"),
    (["cold"], "take_rest"),
]

facts = ["fever", "cough", "body_pain"]

print("Rules:")
for r in rules:
    print("  IF", " AND ".join(r[0]), "THEN", r[1])
print("Facts:", facts)
```

```python
def forward_chaining(rules, facts, goal):
    known = list(facts)          # copy of facts
    changed = True

    while changed:
        changed = False
        for conditions, conclusion in rules:
            all_true = True
            for c in conditions:
                if c not in known:
                    all_true = False

            if all_true and conclusion not in known:
                known.append(conclusion)
                print("Rule fired:", conditions, "->", conclusion)
                changed = True

    return goal in known
```

```python
print("Goal reached?", forward_chaining(rules, facts, "see_doctor"))
```

```python
def backward_chaining(rules, facts, goal):
    if goal in facts:
        print("Fact found:", goal)
        return True

    for conditions, conclusion in rules:
        if conclusion == goal:
            print("To prove", goal, "we need", conditions)
            all_proved = True
            for c in conditions:
                if not backward_chaining(rules, facts, c):
                    all_proved = False
            if all_proved:
                return True

    return False
```

```python
print("Goal reached?", backward_chaining(rules, facts, "see_doctor"))
```

```python
print("Forward :", forward_chaining(rules, facts, "take_rest"))
print("Backward:", backward_chaining(rules, facts, "take_rest"))
```
